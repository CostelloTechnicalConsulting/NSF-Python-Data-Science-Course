```python
import pandas as pd

data_dir = '../data'

awards_df = pd.read_pickle(f'{data_dir}/pkl_files/awards_data_2010_2025.pkl')
```

```python
# Set the program reference code to search for
pgm_ref_code = '0000'

# Filter awards_df to only rows where 'pgm_ref' contains a dict with 
# 'pgm_ref_code' == pgm_ref_code
search_df = awards_df[
    awards_df['pgm_ref'].apply(
        lambda refs: isinstance(refs, list) and any(
            isinstance(ref, dict) and ref.get('pgm_ref_code') == pgm_ref_code
            for ref in refs
        )
    )
]

# Print the number of matching awards
print(f"Number of awards with pgm_ref_code {pgm_ref_code}: {len(search_df)}")

# Display selected columns from the filtered dataframe
display(search_df[['awd_id', 'awd_titl_txt', 'pgm_ref']])
```
